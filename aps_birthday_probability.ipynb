```python
def p_birthday(n):
    if n>365:
        return 1.0
    x=1.0 #probability not shared
    for i in range(n):
        x=(365.0-i)/365.0
    y=1-x #probability shared
    return y

print(p_birthday(2))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate a range of 'n' values (number of people)
n_values = np.arange(1, 401)  # From 1 to 100 people

# Calculate probabilities for each 'n'
probabilities = [p_birthday(n) for n in n_values]

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(n_values, probabilities, marker='o', linestyle='-', color='skyblue')
plt.title('Birthday Paradox Probability (p_birthday(n))')
plt.xlabel('Number of People (n)')
plt.ylabel('Probability of Shared Birthday')
plt.grid(True)
plt.axhline(y=0.5, color='purple', linestyle='--', label='Probability = 0.5')
plt.legend()
plt.show()
```
