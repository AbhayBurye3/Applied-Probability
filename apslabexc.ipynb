```python
import random

# Toss a fair coin 100 times
n = 100

tosses = [random.choice(["Head", "Tail"]) for _ in range(n)]

# Count heads and tails
heads = tosses.count("Head")
tails = tosses.count("Tail")

# Experimental probabilities
p_head = heads / n
p_tail = tails / n

print("Problem 1: Coin Toss")
print("--------------------")
print("Number of heads:", heads)
print("Number of tails:", tails)
print("Experimental probability of Head:", p_head)
print("Experimental probability of Tail:", p_tail)
```

```python
import random

# Different numbers of trials
trial_sizes = [10, 100, 1000, 10000]

print("Problem 2: Effect of Number of Trials")
print("--------------------------------------")

for n in trial_sizes:

    # Toss coin n times
    tosses = [random.choice(["Head", "Tail"]) for _ in range(n)]

    heads = tosses.count("Head")
    tails = tosses.count("Tail")

    p_head = heads / n
    p_tail = tails / n

    print("\nNumber of trials:", n)
    print("Heads:", heads)
    print("Tails:", tails)
    print("P(Head):", p_head)
    print("P(Tail):", p_tail)
```

```python
import random

# Roll a die 10,000 times
n = 10000

rolls = [random.randint(1, 6) for _ in range(n)]

# Event A: getting 1
A_count = rolls.count(1)

# Event B: getting 6
B_count = rolls.count(6)

# Event A union B: getting 1 or 6
A_union_B_count = A_count + B_count

# Probabilities
P_A = A_count / n
P_B = B_count / n
P_A_union_B = A_union_B_count / n

print("Problem 4: Addition Rule for Disjoint Events")
print("----------------------------------------------")

print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A union B) =", P_A_union_B)

print("\nChecking whether events are disjoint:")
print("A = getting 1")
print("B = getting 6")
print("A and B cannot happen at the same time.")
print("Therefore, A and B are DISJOINT events.")

print("\nVerification:")
print("P(A union B) =", P_A_union_B)
print("P(A) + P(B) =", P_A + P_B)
```

```python
# Problem 5: Addition Rule for Overlapping Events

# Sample space
S = {1, 2, 3, 4, 5, 6}

# Define events
A = {2, 4, 6}       # Even numbers
B = {4, 5, 6}       # Numbers greater than 3

# Find intersection and union
A_intersection_B = A.intersection(B)
A_union_B = A.union(B)

# Calculate probabilities
P_A = len(A) / len(S)
P_B = len(B) / len(S)
P_A_intersection_B = len(A_intersection_B) / len(S)
P_A_union_B = len(A_union_B) / len(S)

print("Problem 5: Addition Rule for Overlapping Events")
print("------------------------------------------------")

print("A =", A)
print("B =", B)

print("A intersection B =", A_intersection_B)
print("A union B =", A_union_B)

print("\nProbabilities:")
print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A intersection B) =", P_A_intersection_B)
print("P(A union B) =", P_A_union_B)

# Addition rule
right_side = P_A + P_B - P_A_intersection_B

print("\nVerification:")
print("P(A union B) =", P_A_union_B)
print("P(A) + P(B) - P(A intersection B) =", right_side)
```

```python
import random

# Number of experiments
n = 10000

head_count = 0
even_count = 0
head_and_even_count = 0

# Perform 10,000 experiments
for i in range(n):

    # Toss a coin
    coin = random.choice(["Head", "Tail"])

    # Roll a die
    die = random.randint(1, 6)

    # Event A: Head
    if coin == "Head":
        head_count += 1

    # Event B: Even number
    if die % 2 == 0:
        even_count += 1

    # Event A intersection B
    if coin == "Head" and die % 2 == 0:
        head_and_even_count += 1

# Calculate probabilities
P_A = head_count / n
P_B = even_count / n
P_A_intersection_B = head_and_even_count / n

print("Problem 6: Multiplication Rule")
print("------------------------------")

print("P(A) - Head =", P_A)
print("P(B) - Even number =", P_B)
print("P(A intersection B) =", P_A_intersection_B)

# Verify multiplication rule
calculated = P_A * P_B

print("\nVerification:")
print("P(A intersection B) =", P_A_intersection_B)
print("P(A) * P(B) =", calculated)

print("\nTheoretical values:")
print("P(Head) = 1/2 =", 1/2)
print("P(Even) = 3/6 =", 3/6)
print("P(Head intersection Even) = 1/4 =", 1/4)
```
