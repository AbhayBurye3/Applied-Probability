```python

import numpy as np

# Generate 10,000 random samples
np.random.seed(42)
data = np.random.randint(1, 101, size=10000)

# Mean and median
mean_np = np.mean(data)
median_np = np.median(data)

# Mode
values, counts = np.unique(data, return_counts=True)
mode_np = values[counts == counts.max()]

# Standard deviation (population)
std_np = np.std(data)

print("Using NumPy")
print("Mean:", mean_np)
print("Median:", median_np)
print("Mode:", mode_np)
print("Standard Deviation:", std_np)

```

```python


import numpy as np

# Generate 10,000 random samples
#np.random.seed(42)
#data = np.random.randint(1, 101, size=10000)
numbers = data.tolist()

# Mean
mean_builtin = sum(numbers) / len(numbers)

# Median
sorted_numbers = sorted(numbers)
n = len(sorted_numbers)

median_builtin = (
    sorted_numbers[n // 2 - 1] +
    sorted_numbers[n // 2]
) / 2

# Mode
frequency = {}
for x in numbers:
    frequency[x] = frequency.get(x, 0) + 1

max_frequency = max(frequency.values())
mode_builtin = [
    x for x, count in frequency.items()
    if count == max_frequency
]

# Population standard deviation
variance = sum(
    (x - mean_builtin) ** 2 for x in numbers
) / len(numbers)

std_builtin = variance ** 0.5

print("Using Built-in Functions")
print("Mean:", mean_builtin)
print("Median:", median_builtin)
print("Mode:", mode_builtin)
print("Standard Deviation:", std_builtin)

```

```python

print("\nComparison")

print("Mean equal:",
      np.isclose(mean_np, mean_builtin))

print("Median equal:",
      np.isclose(median_np, median_builtin))

print("Mode equal:",
      np.array_equal(np.sort(mode_np),
                     np.sort(mode_builtin)))

print("Standard deviation equal:",
      np.isclose(std_np, std_builtin))
```
